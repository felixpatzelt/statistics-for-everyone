```python
precipation = pd.concat([
    pd.read_csv(f"https://opendata.dwd.de/climate_environment/CDC/regional_averages_DE/monthly/precipitation/regional_averages_rr_{str(i).rjust(2, '0')}.txt", sep=";", header=1)
    for i in range(1,13)
]).dropna(axis=1)
```

```python
precipation_long = precipation.melt(id_vars=["Monat","Jahr"], var_name="Region", value_name="Precipation")
precipation_mean = precipation_long.groupby(["Monat","Region"]).mean().reset_index()
precipation_max = precipation_mean.Precipation.max()
```

### Descriptive Statistic(s)

- Making a summary ("statistic") that quantitatively describes data
    - The number of participants in a survey 💯
    - The average height of an adult in a certain region 📏
    - The income distribution in a society 💰

- Using and analysing such summaries.

```python
brush = alt.selection_interval(encodings=['x'])

dropdown = alt.binding_select(options=precipation_mean.Region.unique().tolist(), name="Region: ")
selected = alt.selection_point(
    fields=['Region'],
    bind=dropdown,
    value="Deutschland"
)

base = alt.Chart().add_params(
    selected
).transform_filter(
   selected
)

bars = base.mark_bar().encode(
    x=alt.X('Monat:O'),
    y=alt.Y('mean(Precipation):Q').scale(domain=(0, precipation_max)),
    opacity = alt.when(brush).then(alt.value(1)).otherwise(alt.value(0.7)),
).add_params(
    brush
)


line = base.mark_rule(color='firebrick').encode(
    y='mean(Precipation):Q',
    size=alt.SizeValue(3)
).transform_filter(
    brush
)

from IPython.display import display, HTML
display(HTML("""
<style>
form.vega-bindings {
  position: absolute;
  left: 70px;
  top: -60px;
}
</style>
"""))

alt.layer(bars, line, data=precipation_mean).display(renderer='svg')
```

### Inductive Statistics

- Probabilistic analysis 🎲
- Hypothesis testing 🧪
- Prediction and forecasting 📈
- Classification 🗄️
- Generative AI 🤖
- …

```python
distribution = alt.Chart(
    alt.sequence(-3, 3, .1, as_='x')
).mark_line(color='darkblue').encode(
    x=alt.X('x:Q'),
    y=alt.Y('pdf:Q', title='Probability of x'),
).transform_calculate(
    mean = "0",
    std = "1",
    pdf='0.5*densityNormal(datum.x, "0", "1")'
)
samples = alt.Chart(
    alt.sequence(-3, 3, .1)
).transform_calculate(
    mean = "0",
    std  = "1",
    x='round(2*sampleNormal(+datum.mean,+datum.std))/2'
).transform_aggregate(
    count='count()', groupby=['x']
).transform_joinaggregate(
    total='sum(count)'
).transform_calculate(
    probability="datum.count / datum.total"
).mark_bar(color='lightblue').encode(
    x=alt.X('x:Q'),  # Bin the calculated random sample!
    y=alt.Y('probability:Q')
)

mean = alt.Chart(pd.DataFrame({'x': [0], 'y': [0.2]})).mark_point(size=100, color='red').encode(
    x='x:Q',
    y='y:Q'
)
samples + distribution + mean
```
