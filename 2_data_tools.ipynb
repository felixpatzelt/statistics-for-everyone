## There are vast amounts of tools for working with data

### Any programming language can do
- Especially in research and production Python dominates 🐍
- In statistics & analytics, R is still popular 📊
- JavaScript for purely web-focused use cases 🌐
- Compiled languages are used rather for low-level library development or classic app development
- Many specialised languages exist smaller but active communities

### There are also graphical tools with more limited capabilities
- Business reporting (Tableau, PowerBI, Metbase) 👔
- Operations Dashboarding (Grafana, DataDog)
- Statistics (SPSS)
- …

    - Replaced Matlab in Science & Engineering
    - Became the dominant ecosystem for Data Science AI
- In statistics & analytics, R is still popular
- Low-level code is often written in C/C++/Fortran/Rust but this is hardly ever needed for applied uses
